## Decision tree with missing data

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# Load the training dataset
train_filename = 'train.csv'  # Replace with your actual training file name
train_data = pd.read_csv(train_filename)

# Load the testing dataset
test_filename = 'test.csv'  # Replace with your actual testing file name
test_data = pd.read_csv(test_filename)

# Create a DataFrame for complete data (no nulls in 'num_sold')
train_data = train_data[train_data['num_sold'].notnull()]

# Define features and target variable for training
X_train = train_data[['country', 'store', 'product']]
y_train = train_data['num_sold']

# One-hot encoding for categorical variables in training set
X_train_encoded = pd.get_dummies(X_train, drop_first=True)

# Define features for testing data
X_test = test_data[['country', 'store', 'product']]

# One-hot encoding for categorical variables in testing set
X_test_encoded = pd.get_dummies(X_test, drop_first=True)

# Align the columns of the test set with the training set (to handle any differences in categories)
X_test_encoded = X_test_encoded.reindex(columns=X_train_encoded.columns, fill_value=0)

# Initialize the decision tree regressor
model = DecisionTreeRegressor(random_state=42)

# Fit the model to the training data
model.fit(X_train_encoded, y_train)

# Predict on the test dataset
predictions = model.predict(X_test_encoded)

# Round predictions to ensure they are whole numbers
predictions_rounded = predictions.round().astype(int)

# Create a DataFrame for predictions
results_df = pd.DataFrame({
    'ID': test_data['id'],  # Assuming there's an 'id' column in your test data
    'Predicted_num_sold': predictions_rounded
})

# Save predictions to a new CSV file
results_df.to_csv('predictions.csv', index=False)

print("Predictions saved to predictions.csv")
```

```python
# Load the training dataset
train_filename = 'train.csv'  # Replace with your actual training file name
train_data = pd.read_csv(train_filename)

# Create a DataFrame for complete data (no nulls in 'num_sold')
complete_data = train_data[train_data['num_sold'].notnull()]
```

```python
# Define features and target variable for training
X_complete = complete_data[['country', 'store', 'product']]
y_complete = complete_data['num_sold']
```

```python
# Split the complete data into training and validation sets (80% train, 20% validation)
X_train, X_val, y_train, y_val = train_test_split(X_complete, y_complete, test_size=0.2, random_state=42)
```

```python
# Create a neural network model
model = Sequential()
model.add(Dense(64, activation='relu', input_shape=(X_train.shape[1],)))  # Input layer with 64 neurons
model.add(Dense(32, activation='relu'))  # Hidden layer with 32 neurons
model.add(Dense(1))  # Output layer for regression (single output)
```

```python
model.compile(optimizer='adam', loss='mean_absolute_error')
```

```python
# Preprocess the training data and fit the model
X_train_processed = pipeline.fit_transform(X_train)
model.fit(X_train_processed, y_train, epochs=100, batch_size=32, verbose=1)
```

```python
# Preprocess the validation data and make predictions
X_val_processed = pipeline.transform(X_val)
predictions_val = model.predict(X_val_processed)

# Round predictions to ensure they are whole numbers (if needed)
predictions_rounded = predictions_val.round().astype(int)

# Calculate MAPE using actual values from the validation dataset
mape = mean_absolute_percentage_error(y_val, predictions_rounded)
print(f'Mean Absolute Percentage Error on Validation Data: {mape:.2f}%')
```
